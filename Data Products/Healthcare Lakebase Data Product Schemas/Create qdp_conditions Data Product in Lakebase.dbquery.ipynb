```sql
--SELECT version();
--USE DATABASE qdp_database;

CREATE SCHEMA IF NOT EXISTS qdp_conditions;
SET search_path TO qdp_conditions;


-- metadata table
DROP TABLE IF EXISTS metadata;


CREATE TABLE metadata (
	supplier VARCHAR,
	data_product VARCHAR,
	data_product_version VARCHAR,
	schema_name VARCHAR,
	schema_version VARCHAR,
	schema_variant_name VARCHAR,
	schema_variant_version VARCHAR,
	instance VARCHAR,
	instance_version VARCHAR
 );



INSERT INTO metadata
(
  supplier,
  data_product,
  data_product_version,
  schema_name,
  schema_version,
  schema_variant_name,
  schema_variant_version,
  instance,
  instance_version
)
VALUES (
  'quantexa.com',
  'com.quantexa.qdp.conditions',
  '0.1',
  'conditions.data_vault',
  '0.1',
  'base',
  '0.1',
  'not yet populated with data',
  ''
);




-- -----------------------------------------------------------
-- create hub_condition table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS hub_condition CASCADE;

CREATE TABLE hub_condition (
    condition_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
    condition_entity_id VARCHAR,
  	tennant_id BIGINT NOT NULL DEFAULT 0,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE hub_condition ADD CONSTRAINT pk_condition_hubcondition_conditionid PRIMARY KEY (condition_id);

COMMENT ON COLUMN hub_condition.condition_id IS 'Condition Identity';




-- -----------------------------------------------------------
-- create sat_condition table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_condition;

CREATE TABLE sat_condition (
    sat_condition_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
    condition_id BIGINT NOT NULL,
    load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
    record_source_id BIGINT NOT NULL DEFAULT 1,
	data_source_code VARCHAR,
	data_source_concept_id BIGINT,
	data_source_raw_code VARCHAR,
	data_source_raw_concept_id BIGINT,
	encounter_id BIGINT,
	party_id BIGINT,
	clinical_status_code VARCHAR,
	clinical_status_concept_id BIGINT,
	clinical_status_raw_code VARCHAR,
	clinical_status_raw_concept_id BIGINT,
	verification_status_code VARCHAR,
	verification_status_concept_id BIGINT,
	verification_status_raw_code VARCHAR,
	verification_status_raw_concept_id BIGINT,
	severity_code VARCHAR,
	severity_concept_id BIGINT,
	severity_raw_code VARCHAR,
	severity_raw_concept_id BIGINT,
	condition_code VARCHAR,
	condition_code_concept_id BIGINT,
	condition_code_raw_code VARCHAR,
	condition_code_raw_concept_id BIGINT,
	onset_datetime TIMESTAMP,
	onset_age INT,
	onset_period_start TIMESTAMP,
	onset_period_end TIMESTAMP,
	onset_range_low INT,
	onset_range_high INT,
	onset_string VARCHAR,
	abatement_datetime TIMESTAMP,
	abatement_age INT,
	abatement_period_start TIMESTAMP,
	abatement_period_end TIMESTAMP,
	abatement_range_low INT,
	abatement_range_high INT,
	abatement_string VARCHAR,
	recorded_date TIMESTAMP,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_condition ADD CONSTRAINT pk_condition_satcondition_satconditionid PRIMARY KEY (sat_condition_id);
ALTER TABLE sat_condition ADD CONSTRAINT fk_condition_satcondition_hubcondition_conditionid FOREIGN KEY (condition_id) REFERENCES hub_condition(condition_id);

COMMENT ON COLUMN sat_condition.condition_id IS 'Condition Identity';
COMMENT ON COLUMN sat_condition.sat_condition_id IS 'Condition Identity Sat';
COMMENT ON COLUMN sat_condition.encounter_id IS 'The unique Patient Encounter Identifier for this Condition Record';
COMMENT ON COLUMN sat_condition.party_id IS 'The Patient Identifier';
COMMENT ON COLUMN sat_condition.clinical_status_code IS 'The clinical status of the Condition. E.g. active, recurrence, relapse, inactive, remission, resolved, unknown. See http://terminology.hl7.org/CodeSystem/condition-clinical';
COMMENT ON COLUMN sat_condition.verification_status_code IS 'The verification status of the Condition. E.g. unconfirmed, provisional, differential, confirmed, refuted, entered-in-error. See http://terminology.hl7.org/CodeSystem/condition-ver-status';
COMMENT ON COLUMN sat_condition.severity_code IS 'Subjective severity of condition. E.g. Severe (24484000), Moderate (6736007), Mild (255604002). See https://hl7.org/fhir/R5/valueset-condition-severity.html';
COMMENT ON COLUMN sat_condition.condition_code IS 'Identification of the condition, problem or diagnosis (SNOMED CT coded). See https://hl7.org/fhir/R5/valueset-condition-code.html';
COMMENT ON COLUMN sat_condition.onset_datetime IS 'Estimated or actual date-time the Condition began. See https://build.fhir.org/condition-definitions.html#Condition.onset_x_';
COMMENT ON COLUMN sat_condition.onset_age IS 'Estimated or actual age at which the Condition began';
COMMENT ON COLUMN sat_condition.onset_period_start IS 'Start of the estimated period during which the Condition began';
COMMENT ON COLUMN sat_condition.onset_period_end IS 'End of the estimated period during which the Condition began';
COMMENT ON COLUMN sat_condition.onset_range_low IS 'Low end of the estimated age range at which the Condition began';
COMMENT ON COLUMN sat_condition.onset_range_high IS 'High end of the estimated age range at which the Condition began';
COMMENT ON COLUMN sat_condition.onset_string IS 'Textual description of when the Condition began';
COMMENT ON COLUMN sat_condition.abatement_datetime IS 'Date-time when the Condition resolved or went into remission. See https://build.fhir.org/condition-definitions.html#Condition.abatement_x_';
COMMENT ON COLUMN sat_condition.abatement_age IS 'Age at which the Condition resolved or went into remission';
COMMENT ON COLUMN sat_condition.abatement_period_start IS 'Start of the period during which the Condition resolved or went into remission';
COMMENT ON COLUMN sat_condition.abatement_period_end IS 'End of the period during which the Condition resolved or went into remission';
COMMENT ON COLUMN sat_condition.abatement_range_low IS 'Low end of the age range at which the Condition resolved or went into remission';
COMMENT ON COLUMN sat_condition.abatement_range_high IS 'High end of the age range at which the Condition resolved or went into remission';
COMMENT ON COLUMN sat_condition.abatement_string IS 'Textual description of when the Condition resolved or went into remission';
COMMENT ON COLUMN sat_condition.recorded_date IS 'Date condition was first recorded. See https://build.fhir.org/condition-definitions.html#Condition.recordedDate';



-- -----------------------------------------------------------
-- create sat_condition_body_site table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_condition_body_site;

CREATE TABLE sat_condition_body_site (
	condition_body_site_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
	condition_id BIGINT NOT NULL,
	load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
	record_source_id BIGINT NOT NULL DEFAULT 1,
	body_site_code VARCHAR,
	body_site_concept_id BIGINT,
	body_site_raw_code VARCHAR,
	body_site_raw_concept_id BIGINT,
	rank INT,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_condition_body_site ADD CONSTRAINT pk_condition_satconditionbodysite_conditionbodysiteid PRIMARY KEY (condition_body_site_id);
ALTER TABLE sat_condition_body_site ADD CONSTRAINT fk_condition_satconditionbodysite_hubcondition_conditionid FOREIGN KEY (condition_id) REFERENCES hub_condition(condition_id);

COMMENT ON COLUMN sat_condition_body_site.condition_id IS 'Condition Identity';
COMMENT ON COLUMN sat_condition_body_site.condition_body_site_id IS 'Condition Body Site Identity';
COMMENT ON COLUMN sat_condition_body_site.body_site_code IS 'Anatomical location of the Condition, if relevant (SNOMED CT body site coded). See https://hl7.org/fhir/R5/valueset-body-site.html';
COMMENT ON COLUMN sat_condition_body_site.rank IS 'The recommended order the Condition Body Site Codes should be presented';



-- -----------------------------------------------------------
-- create sat_condition_category table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_condition_category;

CREATE TABLE sat_condition_category (
	condition_category_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
	condition_id BIGINT NOT NULL,
	load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
	record_source_id BIGINT NOT NULL DEFAULT 1,
	category_code VARCHAR,
	category_concept_id BIGINT,
	category_raw_code VARCHAR,
	category_raw_concept_id BIGINT,
	rank INT,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_condition_category ADD CONSTRAINT pk_condition_satconditioncategory_conditioncategoryid PRIMARY KEY (condition_category_id);
ALTER TABLE sat_condition_category ADD CONSTRAINT fk_condition_satconditioncategory_hubcondition_conditionid FOREIGN KEY (condition_id) REFERENCES hub_condition(condition_id);

COMMENT ON COLUMN sat_condition_category.condition_id IS 'Condition Identity';
COMMENT ON COLUMN sat_condition_category.condition_category_id IS 'Condition Category Identity';
COMMENT ON COLUMN sat_condition_category.category_code IS 'Category of the Condition. E.g. problem-list-item, encounter-diagnosis. See http://terminology.hl7.org/CodeSystem/condition-category';
COMMENT ON COLUMN sat_condition_category.rank IS 'The recommended order the Condition Category Codes should be presented';



-- -----------------------------------------------------------
-- create sat_condition_evidence table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_condition_evidence;

CREATE TABLE sat_condition_evidence (
	condition_evidence_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
	condition_id BIGINT NOT NULL,
	load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
	record_source_id BIGINT NOT NULL DEFAULT 1,
	evidence_code VARCHAR,
	evidence_concept_id BIGINT,
	evidence_raw_code VARCHAR,
	evidence_raw_concept_id BIGINT,
	actor_party_id BIGINT,
	rank INT,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_condition_evidence ADD CONSTRAINT pk_condition_satconditionevidence_conditionevidenceid PRIMARY KEY (condition_evidence_id);
ALTER TABLE sat_condition_evidence ADD CONSTRAINT fk_condition_satconditionevidence_hubcondition_conditionid FOREIGN KEY (condition_id) REFERENCES hub_condition(condition_id);

COMMENT ON COLUMN sat_condition_evidence.condition_id IS 'Condition Identity';
COMMENT ON COLUMN sat_condition_evidence.condition_evidence_id IS 'Condition Evidence Identity';
COMMENT ON COLUMN sat_condition_evidence.evidence_code IS 'Supporting evidence for the verification status, coded as clinical findings (SNOMED CT). See https://hl7.org/fhir/R5/valueset-clinical-findings.html';
COMMENT ON COLUMN sat_condition_evidence.actor_party_id IS 'Who or what participated in the activities related to the condition. E.g. Practitioner, Patient, RelatedPerson, Device, Organization, CareTeam';
COMMENT ON COLUMN sat_condition_evidence.rank IS 'The recommended order the Condition Evidence Codes should be presented';



-- -----------------------------------------------------------
-- create sat_condition_notes table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_condition_notes;

CREATE TABLE sat_condition_notes (
	condition_note_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
	condition_id BIGINT NOT NULL,
	load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
	record_source_id BIGINT NOT NULL DEFAULT 1,
	author_reference_party_id BIGINT,
	author_string VARCHAR,
	time TIMESTAMP,
	text TEXT NOT NULL,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_condition_notes ADD CONSTRAINT pk_condition_satconditionnotes_conditionnoteid PRIMARY KEY (condition_note_id);
ALTER TABLE sat_condition_notes ADD CONSTRAINT fk_condition_satconditionnotes_hubcondition_conditionid FOREIGN KEY (condition_id) REFERENCES hub_condition(condition_id);

COMMENT ON COLUMN sat_condition_notes.condition_id IS 'Condition Identity';
COMMENT ON COLUMN sat_condition_notes.condition_note_id IS 'Condition Note Identity';
COMMENT ON COLUMN sat_condition_notes.author_reference_party_id IS 'The clinical Party Identifier that recorded this information. E.g. Practitioner, PractitionerRole, Patient, RelatedPerson, Organization. See https://build.fhir.org/condition-definitions.html#Condition.note';
COMMENT ON COLUMN sat_condition_notes.author_string IS 'The individual that recorded this information (as free text)';
COMMENT ON COLUMN sat_condition_notes.time IS 'When this information was recorded';
COMMENT ON COLUMN sat_condition_notes.text IS 'Clinician entered additional information about the Condition';



-- -----------------------------------------------------------
-- create sat_condition_participant table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_condition_participant;

CREATE TABLE sat_condition_participant (
	condition_participant_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
	condition_id BIGINT NOT NULL,
	load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
	record_source_id BIGINT NOT NULL DEFAULT 1,
	function_code VARCHAR,
	function_concept_id BIGINT,
	function_raw_code VARCHAR,
	function_raw_concept_id BIGINT,
	actor_party_id BIGINT,
	rank INT,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_condition_participant ADD CONSTRAINT pk_condition_satconditionparticipant_conditionparticipantid PRIMARY KEY (condition_participant_id);
ALTER TABLE sat_condition_participant ADD CONSTRAINT fk_condition_satconditionparticipant_hubcondition_conditionid FOREIGN KEY (condition_id) REFERENCES hub_condition(condition_id);

COMMENT ON COLUMN sat_condition_participant.condition_id IS 'Condition Identity';
COMMENT ON COLUMN sat_condition_participant.condition_participant_id IS 'Condition Participant Identity';
COMMENT ON COLUMN sat_condition_participant.function_code IS 'Type of involvement in the condition. E.g. enterer, performer, author, verifier, legal, attester, informant, custodian. See https://hl7.org/fhir/R5/valueset-participation-role-type.html';
COMMENT ON COLUMN sat_condition_participant.actor_party_id IS 'Who or what participated in the activities related to the condition. E.g. Practitioner, Patient, RelatedPerson, Device, Organization, CareTeam';
COMMENT ON COLUMN sat_condition_participant.rank IS 'The recommended order the Condition Participant Function Codes should be presented';



-- -----------------------------------------------------------
-- create sat_condition_stage table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_condition_stage CASCADE;

CREATE TABLE sat_condition_stage (
	condition_stage_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
	condition_id BIGINT NOT NULL,
	load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
	record_source_id BIGINT NOT NULL DEFAULT 1,
	summary_code VARCHAR,
	summary_concept_id BIGINT,
	summary_raw_code VARCHAR,
	summary_raw_concept_id BIGINT,
	type_code VARCHAR,
	type_concept_id BIGINT,
	type_raw_code VARCHAR,
	type_raw_concept_id BIGINT,
	actor_party_id BIGINT,
	rank INT,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_condition_stage ADD CONSTRAINT pk_condition_satconditionstage_conditionstageid PRIMARY KEY (condition_stage_id);
ALTER TABLE sat_condition_stage ADD CONSTRAINT fk_condition_satconditionstage_hubcondition_conditionid FOREIGN KEY (condition_id) REFERENCES hub_condition(condition_id);

COMMENT ON COLUMN sat_condition_stage.condition_id IS 'Condition Identity';
COMMENT ON COLUMN sat_condition_stage.condition_stage_id IS 'Condition Stage Identity';
COMMENT ON COLUMN sat_condition_stage.summary_code IS 'Simple summary of the stage (disease specific, SNOMED CT coded). E.g. Clinical stage IV, Lymphoma stage IIe. See https://hl7.org/fhir/R5/valueset-condition-stage.html';
COMMENT ON COLUMN sat_condition_stage.type_code IS 'Kind of staging. E.g. Pathological staging, Clinical staging, Gleason grading, FIGO staging. See https://hl7.org/fhir/R5/valueset-condition-stage-type.html';
COMMENT ON COLUMN sat_condition_stage.actor_party_id IS 'Who or what participated in the activities related to the condition. E.g. Practitioner, Patient, RelatedPerson, Device, Organization, CareTeam';
COMMENT ON COLUMN sat_condition_stage.rank IS 'The recommended order the Condition Stage entries should be presented';



-- -----------------------------------------------------------
-- create sat_condition_stage_assessment table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_condition_stage_assessment;

CREATE TABLE sat_condition_stage_assessment (
	condition_stage_assessment_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
	condition_stage_id BIGINT NOT NULL,
	load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
	record_source_id BIGINT NOT NULL DEFAULT 1,
	clinical_impression_id BIGINT,
	diagnostic_report_id BIGINT,
	observation_id BIGINT,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_condition_stage_assessment ADD CONSTRAINT pk_condition_satconditionstageassessment_conditionstageassessmentid PRIMARY KEY (condition_stage_assessment_id);
ALTER TABLE sat_condition_stage_assessment ADD CONSTRAINT fk_condition_satconditionstageassessment_satconditionstage_conditionstageid FOREIGN KEY (condition_stage_id) REFERENCES sat_condition_stage(condition_stage_id);

COMMENT ON COLUMN sat_condition_stage_assessment.condition_stage_id IS 'Condition Stage Identity';
COMMENT ON COLUMN sat_condition_stage_assessment.condition_stage_assessment_id IS 'Condition Stage Assessment Identity';
COMMENT ON COLUMN sat_condition_stage_assessment.clinical_impression_id IS 'Reference to a ClinicalImpression resource for formal stage assessment. See https://build.fhir.org/condition-definitions.html#Condition.stage.assessment';
COMMENT ON COLUMN sat_condition_stage_assessment.diagnostic_report_id IS 'Reference to a DiagnosticReport resource for formal stage assessment. See https://build.fhir.org/condition-definitions.html#Condition.stage.assessment';
COMMENT ON COLUMN sat_condition_stage_assessment.observation_id IS 'Reference to an Observation resource for formal stage assessment. See https://build.fhir.org/condition-definitions.html#Condition.stage.assessment';



-- -----------------------------------------------------------
-- create view_conditions view
-- -----------------------------------------------------------
CREATE OR REPLACE VIEW view_conditions AS
SELECT
  h.tennant_id,
  t.tennant_name,
  h.condition_id,
	h.condition_entity_id,
	h.period_start AS hub_period_start,
	h.period_end AS hub_period_end,
  s.sat_condition_id,
  s.encounter_id,
  s.party_id,
  s.clinical_status_code,
  s.clinical_status_concept_id,
  s.clinical_status_raw_code,
  s.clinical_status_raw_concept_id,
  s.verification_status_code,
  s.verification_status_concept_id,
  s.verification_status_raw_code,
  s.verification_status_raw_concept_id,
  s.severity_code,
  s.severity_concept_id,
  s.severity_raw_code,
  s.severity_raw_concept_id,
  s.condition_code,
  s.condition_code_concept_id,
  s.condition_code_raw_code,
  s.condition_code_raw_concept_id,
  s.onset_datetime,
  s.onset_age,
  s.onset_period_start,
  s.onset_period_end,
  s.onset_range_low,
  s.onset_range_high,
  s.onset_string,
  s.abatement_datetime,
  s.abatement_age,
  s.abatement_period_start,
  s.abatement_period_end,
  s.abatement_range_low,
  s.abatement_range_high,
  s.abatement_string,
  s.recorded_date,
	s.period_start,
	s.period_end,
	s.load_datetime,
	s.record_source_id,
	s.data_source_code,
	s.data_source_concept_id,
	s.data_source_raw_code,
	s.data_source_raw_concept_id

    FROM budd.qdp_conditions.hub_condition h
    JOIN budd.qdp_conditions.sat_condition s
      ON h.condition_id = s.condition_id
    JOIN budd.qdp_refcodes.tennant t
      on h.tennant_id = t.tennant_id
;

SELECT * from view_conditions;


-- -----------------------------------------------------------
-- create view_condition_categories view
-- -----------------------------------------------------------
CREATE OR REPLACE VIEW view_condition_categories AS
SELECT
  h.tennant_id,
  t.tennant_name,
  h.condition_id,
	h.condition_entity_id,
  s.clinical_status_code,
  s.verification_status_code,
  s.condition_code,
	cc.condition_category_id,
	cc.category_code,
	cc.category_concept_id,
	cc.category_raw_code,
	cc.category_raw_concept_id,
	cc.rank AS category_rank,
	cc.period_start AS category_period_start,
	cc.period_end AS category_period_end,
	h.period_start AS hub_period_start,
	h.period_end AS hub_period_end,
	s.period_start,
	s.period_end,
	s.load_datetime,
	s.record_source_id,
	s.data_source_code,
	s.data_source_concept_id,
	s.data_source_raw_code,
	s.data_source_raw_concept_id

    FROM budd.qdp_conditions.hub_condition h
    JOIN budd.qdp_conditions.sat_condition s
      ON h.condition_id = s.condition_id
    JOIN budd.qdp_conditions.sat_condition_category cc
      ON h.condition_id = cc.condition_id
    JOIN budd.qdp_refcodes.tennant t
      on h.tennant_id = t.tennant_id
;

SELECT * from view_condition_categories;


-- -----------------------------------------------------------
-- create view_condition_stages view
-- -----------------------------------------------------------
CREATE OR REPLACE VIEW view_condition_stages AS
SELECT
  h.tennant_id,
  t.tennant_name,
  h.condition_id,
	h.condition_entity_id,
  s.clinical_status_code,
  s.verification_status_code,
  s.condition_code,
	cs.condition_stage_id,
	cs.summary_code,
	cs.summary_concept_id,
	cs.summary_raw_code,
	cs.summary_raw_concept_id,
	cs.type_code AS stage_type_code,
	cs.type_concept_id AS stage_type_concept_id,
	cs.type_raw_code AS stage_type_raw_code,
	cs.type_raw_concept_id AS stage_type_raw_concept_id,
	cs.actor_party_id AS stage_actor_party_id,
	cs.rank AS stage_rank,
	cs.period_start AS stage_period_start,
	cs.period_end AS stage_period_end,
	h.period_start AS hub_period_start,
	h.period_end AS hub_period_end,
	s.period_start,
	s.period_end,
	s.load_datetime,
	s.record_source_id,
	s.data_source_code,
	s.data_source_concept_id,
	s.data_source_raw_code,
	s.data_source_raw_concept_id

    FROM budd.qdp_conditions.hub_condition h
    JOIN budd.qdp_conditions.sat_condition s
      ON h.condition_id = s.condition_id
    JOIN budd.qdp_conditions.sat_condition_stage cs
      ON h.condition_id = cs.condition_id
    JOIN budd.qdp_refcodes.tennant t
      on h.tennant_id = t.tennant_id
;

SELECT * from view_condition_stages;


-- -----------------------------------------------------------
-- create view_condition_participants view
-- -----------------------------------------------------------
CREATE OR REPLACE VIEW view_condition_participants AS
SELECT
  h.tennant_id,
  t.tennant_name,
  h.condition_id,
	h.condition_entity_id,
  s.clinical_status_code,
  s.verification_status_code,
  s.condition_code,
	cp.condition_participant_id,
	cp.function_code,
	cp.function_concept_id,
	cp.function_raw_code,
	cp.function_raw_concept_id,
	cp.actor_party_id AS participant_actor_party_id,
	cp.rank AS participant_rank,
	cp.period_start AS participant_period_start,
	cp.period_end AS participant_period_end,
	h.period_start AS hub_period_start,
	h.period_end AS hub_period_end,
	s.period_start,
	s.period_end,
	s.load_datetime,
	s.record_source_id,
	s.data_source_code,
	s.data_source_concept_id,
	s.data_source_raw_code,
	s.data_source_raw_concept_id

    FROM budd.qdp_conditions.hub_condition h
    JOIN budd.qdp_conditions.sat_condition s
      ON h.condition_id = s.condition_id
    JOIN budd.qdp_conditions.sat_condition_participant cp
      ON h.condition_id = cp.condition_id
    JOIN budd.qdp_refcodes.tennant t
      on h.tennant_id = t.tennant_id
;

SELECT * from view_condition_participants;


/**

**/
```
