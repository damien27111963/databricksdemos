```sql
--SELECT version();

CREATE SCHEMA IF NOT EXISTS qdp_refcodes;
SET search_path TO qdp_refcodes;


-- metadata table
DROP TABLE IF EXISTS metadata;


CREATE TABLE metadata ( 
	supplier VARCHAR,
	data_product VARCHAR,
	data_product_version VARCHAR,
	schema_name VARCHAR,
	schema_version VARCHAR,
	schema_variant_name VARCHAR,
	schema_variant_version VARCHAR,
	instance VARCHAR,
	instance_version VARCHAR
 );



INSERT INTO metadata 
(
  supplier, 
  data_product,
  data_product_version,
  schema_name,
  schema_version,
  schema_variant_name,
  schema_variant_version,
  instance,
  instance_version
) 
VALUES (
  'quantexa.com',
  'com.quantexa.qdp.refcodes',
  '0.1',
  'qdp_refcodes',
  '0.1',
  'universal',
  '0.1',
  'not yet populated with data',
  ''
);




-- tennant table
DROP TABLE IF EXISTS tennant;

CREATE TABLE tennant ( 
	tennant_id BIGINT,
	tennant_name VARCHAR,
	network_name VARCHAR
 );

-- Record Source table
DROP TABLE IF EXISTS record_source;

CREATE TABLE record_source ( 
	record_source_id BIGINT,
	domain VARCHAR,
	it_system VARCHAR,
	network_name VARCHAR
 );


-- Populate tennant and record_source tables

INSERT INTO tennant 
(
  tennant_id, 
  tennant_name,
  network_name
) 
VALUES (
  0,
  'NO TENNANT',
  'NO NETWORK'
);


INSERT INTO record_source 
(
  record_source_id, 
  domain,
  it_system,
  network_name
) 
VALUES (
  0,
  'NO TENNANT',
  'NO IT SYSTEM',
  'NO RECORD SOURCE'
);


INSERT INTO tennant 
(
  tennant_id, 
  tennant_name,
  network_name
) 
VALUES (
  1,
  'Healthcare Demo - After Quantexa Processing Demo',
  'Healthcare Demo'
);

INSERT INTO record_source 
(
  record_source_id, 
  domain,
  it_system,
  network_name
) 
VALUES (
  1,
  'com.quantexa.demo/healthcare',
  '',
  'Healthcare Demo - After Quantexa Processing'
);


INSERT INTO tennant 
(
  tennant_id, 
  tennant_name,
  network_name
) 
VALUES (
  2,
  'Healthcare Demo - Before Quantexa Processing Demo',
  'Healthcare Demo'
);

INSERT INTO record_source 
(
  record_source_id, 
  domain,
  it_system,
  network_name
) 
VALUES (
  2,
  'com.quantexa.demo/healthcare',
  '',
  'Healthcare Demo - Before Quantexa Processing'
);

SELECT * FROM tennant;



--
--  OMOP Vocabulary Concept Tables
--

-- Domain table
DROP TABLE IF EXISTS domain;

CREATE TABLE domain (
	domain_id VARCHAR NOT NULL,
	domain_concept_id BIGINT NOT NULL,
	domain_name VARCHAR NOT NULL
);

ALTER TABLE Domain ADD CONSTRAINT pk_conceptAll_domain_domainId PRIMARY KEY (domain_id);



-- Vocabulary table
DROP TABLE IF EXISTS vocabulary;

CREATE TABLE Vocabulary (
	vocabulary_id VARCHAR NOT NULL,
	vocabulary_concept_id BIGINT NOT NULL,
	vocabulary_name VARCHAR NOT NULL,
	vocabulary_reference VARCHAR NOT NULL,
	vocabulary_version VARCHAR NOT NULL
);

ALTER TABLE Vocabulary ADD CONSTRAINT pk_conceptAll_vocabulary_vocabularyId PRIMARY KEY (vocabulary_id);



-- ConceptClass table
DROP TABLE IF EXISTS concept_class;


CREATE TABLE concept_class (
	concept_class_id VARCHAR NOT NULL,
	concept_class_concept_id BIGINT NOT NULL,
	concept_class_name VARCHAR NOT NULL
);

ALTER TABLE concept_class ADD CONSTRAINT pk_conceptAll_conceptClass_conceptClassId PRIMARY KEY (concept_class_id);

-- Concept table
DROP TABLE IF EXISTS concept;

CREATE TABLE concept (
	concept_id BIGINT NOT NULL,
	domain_id VARCHAR NOT NULL,
	vocabulary_id VARCHAR NOT NULL,
	concept_class_id VARCHAR NOT NULL,
	standard_concept VARCHAR NOT NULL,
	concept_code VARCHAR NOT NULL,
  valid_start_date DATE NOT NULL,
  valid_end_date DATE NOT NULL,
  invalid_reason VARCHAR
);

ALTER TABLE concept ADD CONSTRAINT pk_conceptAll_concept_conceptId PRIMARY KEY (concept_id);


-- ConceptSynonym table
DROP TABLE IF EXISTS concept_synonym;

CREATE TABLE concept_synonym (
	concept_id BIGINT NOT NULL,
	language_concept_id BIGINT NOT NULL,
	concept_synonym_name VARCHAR NOT NULL
);


-- Relationship table
DROP TABLE IF EXISTS relationship;

CREATE TABLE relationship (
	relationship_id VARCHAR NOT NULL,
	relationship_concept_id BIGINT NOT NULL,
  reverse_relationship_id VARCHAR NOT NULL,
	relationship_name VARCHAR NOT NULL,
  is_hierarchical BOOLEAN NOT NULL DEFAULT FALSE,
  defines_ancestry BOOLEAN NOT NULL DEFAULT FALSE
);

ALTER TABLE relationship ADD CONSTRAINT pk_conceptAll_relationship_relationshipId PRIMARY KEY (relationship_id);


-- Concpet Relationship table
DROP TABLE IF EXISTS concept_relationship;

CREATE TABLE concept_relationship (
	concept_id1 BIGINT NOT NULL,
	concept_id2 BIGINT NOT NULL,
	relationship_id VARCHAR NOT NULL,
  valid_start_date DATE NOT NULL,
  valid_end_date DATE NOT NULL,
  invalid_reason VARCHAR
);

ALTER TABLE concept_relationship ADD CONSTRAINT pk_conceptAll_ConceptRelationship_conceptId1_conceptID2 PRIMARY KEY (concept_id1, concept_id2);

```
