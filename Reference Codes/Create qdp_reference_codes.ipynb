```sql
-- Create Quantexa Data Product concepts_all Schema
USE CATALOG IDENTIFIER(:p_catalog_name);
USE SCHEMA IDENTIFIER(:p_schema_name);

-- Domain table
DROP TABLE IF EXISTS domain;

CREATE OR REPLACE TABLE domain (
	domain_id STRING NOT NULL,
	domain_concept_id BIGINT NOT NULL,
	domain_name STRING NOT NULL
) USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported');

ALTER TABLE domain ADD CONSTRAINT pk_conceptAll_domain_domainId PRIMARY KEY (domain_id);

```

```sql
-- Create Quantexa Data Product concepts_all Schema

-- Vocabulary table
DROP TABLE IF EXISTS vocabulary;

CREATE OR REPLACE TABLE vocabulary (
	vocabulary_id STRING NOT NULL,
	vocabulary_concept_id BIGINT NOT NULL,
	vocabulary_name STRING NOT NULL,
	vocabulary_reference STRING NOT NULL,
	vocabulary_version STRING NOT NULL
) USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported');

ALTER TABLE vocabulary ADD CONSTRAINT pk_conceptAll_vocabulary_vocabularyId PRIMARY KEY (vocabulary_id);
```

```sql
-- Create Quantexa Data Product concepts_aclass Schema

-- ConceptClass table
DROP TABLE IF EXISTS concept_class;

CREATE OR REPLACE TABLE concept_class (
	concept_class_id STRING NOT NULL,
	concept_class_concept_id BIGINT NOT NULL,
	concept_class_name STRING NOT NULL
) USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported');

ALTER TABLE concept_class ADD CONSTRAINT pk_conceptAll_conceptClass_conceptClassId PRIMARY KEY (concept_class_id);
```

```sql
-- Create Quantexa Data Product concepts_all Schema

-- Concept table
DROP TABLE IF EXISTS concept;

CREATE OR REPLACE TABLE concept (
	concept_id BIGINT NOT NULL,
	domain_id STRING NOT NULL,
	vocabulary_id STRING NOT NULL,
	concept_class_id STRING NOT NULL,
	standard_concept STRING NOT NULL,
	concept_code STRING NOT NULL,
    valid_start_date DATE NOT NULL,
    valid_end_date DATE NOT NULL,
    invalid_reason STRING
) USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported');

ALTER TABLE Concept ADD CONSTRAINT pk_conceptAll_concept_conceptId PRIMARY KEY (concept_id);
```

```sql
-- Create Quantexa Data Product concepts_all Schema

-- ConceptSynonym table
DROP TABLE IF EXISTS concept_synonym;

CREATE OR REPLACE TABLE concept_synonym (
	concept_id BIGINT NOT NULL,
	language_concept_id BIGINT NOT NULL,
	concept_synonym_name STRING NOT NULL
) USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported');


```

```sql
-- Create Quantexa Data Product concepts_all Schema

-- Relationship table
DROP TABLE IF EXISTS relationship;

CREATE OR REPLACE TABLE relationship (
	relationship_id STRING NOT NULL,
	relationship_concept_id BIGINT NOT NULL,
  reverse_relationship_id STRING NOT NULL,
	relationship_name STRING NOT NULL,
  is_hierarchical BOOLEAN NOT NULL DEFAULT FALSE,
  defines_ancestry BOOLEAN NOT NULL DEFAULT FALSE
) USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported');

ALTER TABLE Relationship ADD CONSTRAINT pk_conceptAll_relationship_relationshipId PRIMARY KEY (relationship_id);
```

```sql
-- Create Quantexa Data Product concept_relationships Schema

-- Concept Relationship table
DROP TABLE IF EXISTS concept_relationship;

CREATE OR REPLACE TABLE concept_relationship (
	concept_id1 BIGINT NOT NULL,
	concept_id2 BIGINT NOT NULL,
	relationship_id STRING NOT NULL,
  valid_start_date DATE NOT NULL,
  valid_end_date DATE NOT NULL,
  invalid_reason STRING
) USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported');

ALTER TABLE concept_relationship ADD CONSTRAINT pk_conceptAll_ConceptRelationship_conceptId1_conceptId2 PRIMARY KEY (concept_id1, concept_id2);
```
