```sql
--SELECT version();
--USE DATABASE qdp_database;

CREATE SCHEMA IF NOT EXISTS qdp_immunizations;
SET search_path TO qdp_immunizations;


-- metadata table
DROP TABLE IF EXISTS metadata;


CREATE TABLE metadata (
	supplier VARCHAR,
	data_product VARCHAR,
	data_product_version VARCHAR,
	schema_name VARCHAR,
	schema_version VARCHAR,
	schema_variant_name VARCHAR,
	schema_variant_version VARCHAR,
	instance VARCHAR,
	instance_version VARCHAR
 );



INSERT INTO metadata
(
  supplier,
  data_product,
  data_product_version,
  schema_name,
  schema_version,
  schema_variant_name,
  schema_variant_version,
  instance,
  instance_version
)
VALUES (
  'quantexa.com',
  'com.quantexa.qdp.immunizations',
  '0.1',
  'immunizations.data_vault',
  '0.1',
  'base',
  '0.1',
  'not yet populated with data',
  ''
);




-- -----------------------------------------------------------
-- create hub_immunization table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS hub_immunization CASCADE;

CREATE TABLE hub_immunization (
    immunization_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
    immunization_entity_id VARCHAR,
  	tennant_id BIGINT NOT NULL DEFAULT 0,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE hub_immunization ADD CONSTRAINT pk_immunization_hubimmunization_immunizationid PRIMARY KEY (immunization_id);

COMMENT ON COLUMN hub_immunization.immunization_id IS 'Immunization Identity';




-- -----------------------------------------------------------
-- create sat_immunization table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_immunization;

CREATE TABLE sat_immunization (
    sat_immunization_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
    immunization_id BIGINT NOT NULL,
    load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
    record_source_id BIGINT NOT NULL DEFAULT 1,
	data_source_code VARCHAR,
	data_source_concept_id BIGINT,
	data_source_raw_code VARCHAR,
	data_source_raw_concept_id BIGINT,
	encounter_id BIGINT,
	party_id BIGINT,
	status_code VARCHAR,
	status_concept_id BIGINT,
	status_raw_code VARCHAR,
	status_raw_concept_id BIGINT,
	status_reason_code VARCHAR,
	status_reason_concept_id BIGINT,
	status_reason_raw_code VARCHAR,
	status_reason_raw_concept_id BIGINT,
	vaccine_code VARCHAR,
	vaccine_concept_id BIGINT,
	vaccine_raw_code VARCHAR,
	vaccine_raw_concept_id BIGINT,
	administered_product_code VARCHAR,
	administered_product_concept_id BIGINT,
	administered_product_raw_code VARCHAR,
	administered_product_raw_concept_id BIGINT,
	manufacturer VARCHAR,
	manufacturer_organisation_id BIGINT,
	lot_number VARCHAR,
	expiration_date DATE,
	occurrence_datetime TIMESTAMP,
	occurrence_string VARCHAR,
	primary_source BOOLEAN,
	information_source_code VARCHAR,
	information_source_concept_id BIGINT,
	information_source_raw_code VARCHAR,
	information_source_raw_concept_id BIGINT,
	location_id BIGINT,
	site_code VARCHAR,
	site_concept_id BIGINT,
	site_raw_code VARCHAR,
	site_raw_concept_id BIGINT,
	route_code VARCHAR,
	route_concept_id BIGINT,
	route_raw_code VARCHAR,
	route_raw_concept_id BIGINT,
	dose_quantity_value DOUBLE PRECISION,
	dose_quantity_unit VARCHAR,
	dose_quantity_unit_code VARCHAR,
	is_subpotent BOOLEAN,
	funding_source_code VARCHAR,
	funding_source_concept_id BIGINT,
	funding_source_raw_code VARCHAR,
	funding_source_raw_concept_id BIGINT,
	record_id BIGINT,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_immunization ADD CONSTRAINT pk_immunization_satimmunization_satimmunizationid PRIMARY KEY (sat_immunization_id);
ALTER TABLE sat_immunization ADD CONSTRAINT fk_immunization_satimmunization_hubimmunization_immunizationid FOREIGN KEY (immunization_id) REFERENCES hub_immunization(immunization_id);

COMMENT ON COLUMN sat_immunization.immunization_id IS 'Immunization Identity';
COMMENT ON COLUMN sat_immunization.sat_immunization_id IS 'Immunization Identity Sat';
COMMENT ON COLUMN sat_immunization.encounter_id IS 'The visit or admission or other contact between patient and health care provider the immunization was performed as part of. See https://build.fhir.org/immunization-definitions.html#Immunization.encounter';
COMMENT ON COLUMN sat_immunization.party_id IS 'The Patient Identifier. The patient who either received or did not receive the immunization. See https://build.fhir.org/immunization-definitions.html#Immunization.patient';
COMMENT ON COLUMN sat_immunization.status_code IS 'Status of the Immunization event. E.g. completed, entered-in-error, not-done. See https://hl7.org/fhir/R5/valueset-immunization-status.html';
COMMENT ON COLUMN sat_immunization.status_reason_code IS 'Indicates the reason the immunization event was not performed. E.g. immunity, medical-precaution, product-unavailable, patient-objection. See https://hl7.org/fhir/R5/valueset-immunization-status-reason.html';
COMMENT ON COLUMN sat_immunization.vaccine_code IS 'Vaccine product administered. CVX (vaccine administered) coded. See https://build.fhir.org/immunization-definitions.html#Immunization.vaccineCode';
COMMENT ON COLUMN sat_immunization.administered_product_code IS 'The product that was administered. An actual product administered to the patient. See https://build.fhir.org/immunization-definitions.html#Immunization.administeredProduct';
COMMENT ON COLUMN sat_immunization.manufacturer IS 'Name of vaccine manufacturer. See https://build.fhir.org/immunization-definitions.html#Immunization.manufacturer';
COMMENT ON COLUMN sat_immunization.manufacturer_organisation_id IS 'Organisation reference for the vaccine manufacturer';
COMMENT ON COLUMN sat_immunization.lot_number IS 'Lot number of the vaccine product. See https://build.fhir.org/immunization-definitions.html#Immunization.lotNumber';
COMMENT ON COLUMN sat_immunization.expiration_date IS 'Date vaccine batch expires. See https://build.fhir.org/immunization-definitions.html#Immunization.expirationDate';
COMMENT ON COLUMN sat_immunization.occurrence_datetime IS 'Date vaccine administered or was to be administered. See https://build.fhir.org/immunization-definitions.html#Immunization.occurrence_x_';
COMMENT ON COLUMN sat_immunization.occurrence_string IS 'Vaccine administration date expressed as text when a precise date is not available';
COMMENT ON COLUMN sat_immunization.primary_source IS 'Indicates whether the data contained in the resource was captured by the individual/organization which was responsible for the administration of the vaccine. See https://build.fhir.org/immunization-definitions.html#Immunization.primarySource';
COMMENT ON COLUMN sat_immunization.information_source_code IS 'Indicates the source of a reported record. E.g. newborn, recall, school, provider. See https://hl7.org/fhir/R5/valueset-immunization-origin.html';
COMMENT ON COLUMN sat_immunization.location_id IS 'The service delivery location where the vaccine administration occurred. See https://build.fhir.org/immunization-definitions.html#Immunization.location';
COMMENT ON COLUMN sat_immunization.site_code IS 'Body site where vaccine was administered. E.g. left arm, right arm, left thigh, right thigh. See https://hl7.org/fhir/R5/valueset-immunization-site.html';
COMMENT ON COLUMN sat_immunization.route_code IS 'The path by which the vaccine product is taken into the body. E.g. IM (intramuscular), PO (oral), SC (subcutaneous), ID (intradermal). See https://hl7.org/fhir/R5/valueset-immunization-route.html';
COMMENT ON COLUMN sat_immunization.dose_quantity_value IS 'The quantity of vaccine product that was administered. See https://build.fhir.org/immunization-definitions.html#Immunization.doseQuantity';
COMMENT ON COLUMN sat_immunization.dose_quantity_unit IS 'The unit of measure for the dose quantity (display text)';
COMMENT ON COLUMN sat_immunization.dose_quantity_unit_code IS 'The coded unit of measure for the dose quantity (UCUM code)';
COMMENT ON COLUMN sat_immunization.is_subpotent IS 'Indicates whether the dose is subpotent. By default, a dose should be considered to be potent. See https://build.fhir.org/immunization-definitions.html#Immunization.isSubpotent';
COMMENT ON COLUMN sat_immunization.funding_source_code IS 'Indicates the source of the vaccine actually administered. This may be different than the patient eligibility (e.g. the patient may be eligible for a publically purchased vaccine but due to inventory issues, was administered a privately purchased one). See https://hl7.org/fhir/R5/valueset-immunization-funding-source.html';
COMMENT ON COLUMN sat_immunization.record_id IS 'A Unique Identifier for a Data Record within the System that allows Data Ownership, Ownership History, Dynamic Attributes and Metadata to be associated with that Record';



-- -----------------------------------------------------------
-- create sat_immunization_performer table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_immunization_performer;

CREATE TABLE sat_immunization_performer (
	immunization_performer_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
	immunization_id BIGINT NOT NULL,
	load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
	record_source_id BIGINT NOT NULL DEFAULT 1,
	performer_party_id BIGINT,
	function_code VARCHAR,
	function_concept_id BIGINT,
	function_raw_code VARCHAR,
	function_raw_concept_id BIGINT,
	rank INT,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_immunization_performer ADD CONSTRAINT pk_immunization_satimmperf_immperfid PRIMARY KEY (immunization_performer_id);
ALTER TABLE sat_immunization_performer ADD CONSTRAINT fk_immunization_satimmperf_hubimmunization_immunizationid FOREIGN KEY (immunization_id) REFERENCES hub_immunization(immunization_id);

COMMENT ON COLUMN sat_immunization_performer.immunization_id IS 'Immunization Identity';
COMMENT ON COLUMN sat_immunization_performer.immunization_performer_id IS 'Immunization Performer Identity';
COMMENT ON COLUMN sat_immunization_performer.performer_party_id IS 'The practitioner or organization who performed the action. E.g. Practitioner, PractitionerRole, Organization, Patient, RelatedPerson. See https://build.fhir.org/immunization-definitions.html#Immunization.performer.actor';
COMMENT ON COLUMN sat_immunization_performer.function_code IS 'Describes the type of performance. E.g. ordering provider (OP), administering provider (AP). See https://hl7.org/fhir/R5/valueset-immunization-function.html';
COMMENT ON COLUMN sat_immunization_performer.rank IS 'The recommended order the Immunization Performer entries should be presented';



-- -----------------------------------------------------------
-- create sat_immunization_reason table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_immunization_reason;

CREATE TABLE sat_immunization_reason (
	immunization_reason_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
	immunization_id BIGINT NOT NULL,
	load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
	record_source_id BIGINT NOT NULL DEFAULT 1,
	reason_code VARCHAR,
	reason_concept_id BIGINT,
	reason_raw_code VARCHAR,
	reason_raw_concept_id BIGINT,
	reason_reference_type VARCHAR,
	reason_reference_id BIGINT,
	rank INT,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_immunization_reason ADD CONSTRAINT pk_immunization_satimmreason_immreasonid PRIMARY KEY (immunization_reason_id);
ALTER TABLE sat_immunization_reason ADD CONSTRAINT fk_immunization_satimmreason_hubimmunization_immunizationid FOREIGN KEY (immunization_id) REFERENCES hub_immunization(immunization_id);

COMMENT ON COLUMN sat_immunization_reason.immunization_id IS 'Immunization Identity';
COMMENT ON COLUMN sat_immunization_reason.immunization_reason_id IS 'Immunization Reason Identity';
COMMENT ON COLUMN sat_immunization_reason.reason_code IS 'Indicates why the immunization occurred. E.g. SNOMED CT clinical finding or procedure codes. See https://build.fhir.org/immunization-definitions.html#Immunization.reason';
COMMENT ON COLUMN sat_immunization_reason.reason_reference_type IS 'Type of the reason reference resource (e.g. Condition, Observation, DiagnosticReport)';
COMMENT ON COLUMN sat_immunization_reason.reason_reference_id IS 'Identifier of the reason reference resource';
COMMENT ON COLUMN sat_immunization_reason.rank IS 'The recommended order the Immunization Reason entries should be presented';



-- -----------------------------------------------------------
-- create sat_immunization_reaction table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_immunization_reaction;

CREATE TABLE sat_immunization_reaction (
	immunization_reaction_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
	immunization_id BIGINT NOT NULL,
	load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
	record_source_id BIGINT NOT NULL DEFAULT 1,
	reaction_date TIMESTAMP,
	manifestation_code VARCHAR,
	manifestation_concept_id BIGINT,
	manifestation_raw_code VARCHAR,
	manifestation_raw_concept_id BIGINT,
	manifestation_observation_id BIGINT,
	reported BOOLEAN,
	rank INT,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_immunization_reaction ADD CONSTRAINT pk_immunization_satimmreaction_immreactionid PRIMARY KEY (immunization_reaction_id);
ALTER TABLE sat_immunization_reaction ADD CONSTRAINT fk_immunization_satimmreaction_hubimmunization_immunizationid FOREIGN KEY (immunization_id) REFERENCES hub_immunization(immunization_id);

COMMENT ON COLUMN sat_immunization_reaction.immunization_id IS 'Immunization Identity';
COMMENT ON COLUMN sat_immunization_reaction.immunization_reaction_id IS 'Immunization Reaction Identity';
COMMENT ON COLUMN sat_immunization_reaction.reaction_date IS 'Date of reaction to the immunization. See https://build.fhir.org/immunization-definitions.html#Immunization.reaction.date';
COMMENT ON COLUMN sat_immunization_reaction.manifestation_code IS 'Details of the reaction as a coded concept. SNOMED CT Clinical Finding coded. See https://build.fhir.org/immunization-definitions.html#Immunization.reaction.manifestation';
COMMENT ON COLUMN sat_immunization_reaction.manifestation_observation_id IS 'Cross-schema reference to an Observation that provides additional details of the reaction';
COMMENT ON COLUMN sat_immunization_reaction.reported IS 'Self-reported indicator. TRUE indicates the reaction was self-reported by the patient or caregiver. See https://build.fhir.org/immunization-definitions.html#Immunization.reaction.reported';
COMMENT ON COLUMN sat_immunization_reaction.rank IS 'The recommended order the Immunization Reaction entries should be presented';



-- -----------------------------------------------------------
-- create sat_immunization_protocol_applied table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_immunization_protocol_applied;

CREATE TABLE sat_immunization_protocol_applied (
	immunization_protocol_applied_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
	immunization_id BIGINT NOT NULL,
	load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
	record_source_id BIGINT NOT NULL DEFAULT 1,
	series VARCHAR,
	authority_organisation_id BIGINT,
	target_disease_code VARCHAR,
	target_disease_concept_id BIGINT,
	target_disease_raw_code VARCHAR,
	target_disease_raw_concept_id BIGINT,
	dose_number VARCHAR,
	series_doses VARCHAR,
	rank INT,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_immunization_protocol_applied ADD CONSTRAINT pk_immunization_satimmprotocol_immprotocolid PRIMARY KEY (immunization_protocol_applied_id);
ALTER TABLE sat_immunization_protocol_applied ADD CONSTRAINT fk_immunization_satimmprotocol_hubimmunization_immunizationid FOREIGN KEY (immunization_id) REFERENCES hub_immunization(immunization_id);

COMMENT ON COLUMN sat_immunization_protocol_applied.immunization_id IS 'Immunization Identity';
COMMENT ON COLUMN sat_immunization_protocol_applied.immunization_protocol_applied_id IS 'Immunization Protocol Applied Identity';
COMMENT ON COLUMN sat_immunization_protocol_applied.series IS 'One possible path to achieve presumed immunity against a disease — within the context of an authority. E.g. the name of the vaccine series. See https://build.fhir.org/immunization-definitions.html#Immunization.protocolApplied.series';
COMMENT ON COLUMN sat_immunization_protocol_applied.authority_organisation_id IS 'Indicates the authority who published the protocol (e.g. ACIP, JCVI) followed. See https://build.fhir.org/immunization-definitions.html#Immunization.protocolApplied.authority';
COMMENT ON COLUMN sat_immunization_protocol_applied.target_disease_code IS 'The vaccine preventable disease the dose is being administered against. SNOMED CT coded. See https://build.fhir.org/immunization-definitions.html#Immunization.protocolApplied.targetDisease';
COMMENT ON COLUMN sat_immunization_protocol_applied.dose_number IS 'Nominal position of the dose in a series as determined by the outcome of the evaluation process. E.g. 1, 2, 3 or dose (when series is unknown). See https://build.fhir.org/immunization-definitions.html#Immunization.protocolApplied.doseNumber';
COMMENT ON COLUMN sat_immunization_protocol_applied.series_doses IS 'The recommended number of doses to achieve immunity as determined by the outcome of the evaluation process. E.g. 2, 3. See https://build.fhir.org/immunization-definitions.html#Immunization.protocolApplied.seriesDoses';
COMMENT ON COLUMN sat_immunization_protocol_applied.rank IS 'The recommended order the Immunization Protocol Applied entries should be presented';



-- -----------------------------------------------------------
-- create sat_immunization_program_eligibility table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_immunization_program_eligibility;

CREATE TABLE sat_immunization_program_eligibility (
	immunization_program_eligibility_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
	immunization_id BIGINT NOT NULL,
	load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
	record_source_id BIGINT NOT NULL DEFAULT 1,
	program_code VARCHAR,
	program_concept_id BIGINT,
	program_raw_code VARCHAR,
	program_raw_concept_id BIGINT,
	program_status_code VARCHAR,
	program_status_concept_id BIGINT,
	program_status_raw_code VARCHAR,
	program_status_raw_concept_id BIGINT,
	rank INT,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_immunization_program_eligibility ADD CONSTRAINT pk_immunization_satimmprogram_immprogramid PRIMARY KEY (immunization_program_eligibility_id);
ALTER TABLE sat_immunization_program_eligibility ADD CONSTRAINT fk_immunization_satimmprogram_hubimmunization_immunizationid FOREIGN KEY (immunization_id) REFERENCES hub_immunization(immunization_id);

COMMENT ON COLUMN sat_immunization_program_eligibility.immunization_id IS 'Immunization Identity';
COMMENT ON COLUMN sat_immunization_program_eligibility.immunization_program_eligibility_id IS 'Immunization Program Eligibility Identity';
COMMENT ON COLUMN sat_immunization_program_eligibility.program_code IS 'Indicates which program the patient had their eligibility evaluated for. E.g. a national vaccination program (UK COVID-19 Vaccination Programme, NHS Flu Vaccination Programme). See https://build.fhir.org/immunization-definitions.html#Immunization.programEligibility.program';
COMMENT ON COLUMN sat_immunization_program_eligibility.program_status_code IS 'Indicates the patient''s eligibility status for the specific vaccination program. E.g. ineligible, uninsured. See https://hl7.org/fhir/R5/valueset-immunization-program-eligibility.html';
COMMENT ON COLUMN sat_immunization_program_eligibility.rank IS 'The recommended order the Immunization Program Eligibility entries should be presented';



-- -----------------------------------------------------------
-- create sat_immunization_subpotent_reason table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_immunization_subpotent_reason;

CREATE TABLE sat_immunization_subpotent_reason (
	immunization_subpotent_reason_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
	immunization_id BIGINT NOT NULL,
	load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
	record_source_id BIGINT NOT NULL DEFAULT 1,
	subpotent_reason_code VARCHAR,
	subpotent_reason_concept_id BIGINT,
	subpotent_reason_raw_code VARCHAR,
	subpotent_reason_raw_concept_id BIGINT,
	rank INT,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_immunization_subpotent_reason ADD CONSTRAINT pk_immunization_satimmsubpotent_immsubpotentid PRIMARY KEY (immunization_subpotent_reason_id);
ALTER TABLE sat_immunization_subpotent_reason ADD CONSTRAINT fk_immunization_satimmsubpotent_hubimmunization_immunizationid FOREIGN KEY (immunization_id) REFERENCES hub_immunization(immunization_id);

COMMENT ON COLUMN sat_immunization_subpotent_reason.immunization_id IS 'Immunization Identity';
COMMENT ON COLUMN sat_immunization_subpotent_reason.immunization_subpotent_reason_id IS 'Immunization Subpotent Reason Identity';
COMMENT ON COLUMN sat_immunization_subpotent_reason.subpotent_reason_code IS 'Reason why a dose is considered to be subpotent. E.g. partial-dose, cold-chain-break, recall, adversestorage, expired. See https://hl7.org/fhir/R5/valueset-immunization-subpotent-reason.html';
COMMENT ON COLUMN sat_immunization_subpotent_reason.rank IS 'The recommended order the Immunization Subpotent Reason entries should be presented';



-- -----------------------------------------------------------
-- create sat_immunization_notes table
-- -----------------------------------------------------------
DROP TABLE IF EXISTS sat_immunization_notes;

CREATE TABLE sat_immunization_notes (
	immunization_note_id BIGINT NOT NULL GENERATED ALWAYS AS IDENTITY (MINVALUE 0 START WITH 0 INCREMENT BY 1),
	immunization_id BIGINT NOT NULL,
	load_datetime TIMESTAMP NOT NULL DEFAULT CURRENT_TIMESTAMP,
	record_source_id BIGINT NOT NULL DEFAULT 1,
	author_reference_party_id BIGINT,
	author_string VARCHAR,
	time TIMESTAMP,
	text TEXT NOT NULL,
	period_start TIMESTAMP,
	period_end TIMESTAMP
);

ALTER TABLE sat_immunization_notes ADD CONSTRAINT pk_immunization_satimmnotes_immnoteid PRIMARY KEY (immunization_note_id);
ALTER TABLE sat_immunization_notes ADD CONSTRAINT fk_immunization_satimmnotes_hubimmunization_immunizationid FOREIGN KEY (immunization_id) REFERENCES hub_immunization(immunization_id);

COMMENT ON COLUMN sat_immunization_notes.immunization_id IS 'Immunization Identity';
COMMENT ON COLUMN sat_immunization_notes.immunization_note_id IS 'Immunization Note Identity';
COMMENT ON COLUMN sat_immunization_notes.author_reference_party_id IS 'The clinical Party Identifier that recorded this annotation. E.g. Practitioner, PractitionerRole, Patient, RelatedPerson, Organization. See https://build.fhir.org/immunization-definitions.html#Immunization.note';
COMMENT ON COLUMN sat_immunization_notes.author_string IS 'The Individual that recorded this annotation (as free text)';
COMMENT ON COLUMN sat_immunization_notes.time IS 'When this annotation was recorded';
COMMENT ON COLUMN sat_immunization_notes.text IS 'Extra information about the immunization that is not conveyed by the other attributes. See https://build.fhir.org/immunization-definitions.html#Immunization.note';




-- -----------------------------------------------------------
-- create view_immunizations view
-- -----------------------------------------------------------
CREATE OR REPLACE VIEW view_immunizations AS
SELECT
  h.tennant_id,
  t.tennant_name,
  h.immunization_id,
	h.immunization_entity_id,
	h.period_start AS hub_period_start,
	h.period_end AS hub_period_end,
  s.sat_immunization_id,
  s.encounter_id,
  s.party_id,
  s.status_code,
  s.status_concept_id,
  s.status_raw_code,
  s.status_raw_concept_id,
  s.status_reason_code,
  s.status_reason_concept_id,
  s.status_reason_raw_code,
  s.status_reason_raw_concept_id,
  s.vaccine_code,
  s.vaccine_concept_id,
  s.vaccine_raw_code,
  s.vaccine_raw_concept_id,
  s.administered_product_code,
  s.administered_product_concept_id,
  s.administered_product_raw_code,
  s.administered_product_raw_concept_id,
  s.manufacturer,
  s.manufacturer_organisation_id,
  s.lot_number,
  s.expiration_date,
  s.occurrence_datetime,
  s.occurrence_string,
  s.primary_source,
  s.information_source_code,
  s.information_source_concept_id,
  s.information_source_raw_code,
  s.information_source_raw_concept_id,
  s.location_id,
  s.site_code,
  s.site_concept_id,
  s.site_raw_code,
  s.site_raw_concept_id,
  s.route_code,
  s.route_concept_id,
  s.route_raw_code,
  s.route_raw_concept_id,
  s.dose_quantity_value,
  s.dose_quantity_unit,
  s.dose_quantity_unit_code,
  s.is_subpotent,
  s.funding_source_code,
  s.funding_source_concept_id,
  s.funding_source_raw_code,
  s.funding_source_raw_concept_id,
  s.record_id,
	s.period_start,
	s.period_end,
	s.load_datetime,
	s.record_source_id,
	s.data_source_code,
	s.data_source_concept_id,
	s.data_source_raw_code,
	s.data_source_raw_concept_id

    FROM budd.qdp_immunizations.hub_immunization h
    JOIN budd.qdp_immunizations.sat_immunization s
      ON h.immunization_id = s.immunization_id
    JOIN budd.qdp_refcodes.tennant t
      on h.tennant_id = t.tennant_id
;

SELECT * from view_immunizations;


-- -----------------------------------------------------------
-- create view_immunization_performers view
-- -----------------------------------------------------------
CREATE OR REPLACE VIEW view_immunization_performers AS
SELECT
  h.tennant_id,
  t.tennant_name,
  h.immunization_id,
	h.immunization_entity_id,
  s.vaccine_code,
  s.vaccine_concept_id,
  s.status_code,
  s.occurrence_datetime,
	ip.immunization_performer_id,
	ip.performer_party_id,
	ip.function_code,
	ip.function_concept_id,
	ip.function_raw_code,
	ip.function_raw_concept_id,
	ip.rank AS performer_rank,
	ip.period_start AS performer_period_start,
	ip.period_end AS performer_period_end,
	h.period_start AS hub_period_start,
	h.period_end AS hub_period_end,
	s.period_start,
	s.period_end,
	s.load_datetime,
	s.record_source_id,
	s.data_source_code,
	s.data_source_concept_id,
	s.data_source_raw_code,
	s.data_source_raw_concept_id

    FROM budd.qdp_immunizations.hub_immunization h
    JOIN budd.qdp_immunizations.sat_immunization s
      ON h.immunization_id = s.immunization_id
    JOIN budd.qdp_immunizations.sat_immunization_performer ip
      ON h.immunization_id = ip.immunization_id
    JOIN budd.qdp_refcodes.tennant t
      on h.tennant_id = t.tennant_id
;

SELECT * from view_immunization_performers;


-- -----------------------------------------------------------
-- create view_immunization_reactions view
-- -----------------------------------------------------------
CREATE OR REPLACE VIEW view_immunization_reactions AS
SELECT
  h.tennant_id,
  t.tennant_name,
  h.immunization_id,
	h.immunization_entity_id,
  s.vaccine_code,
  s.vaccine_concept_id,
  s.status_code,
  s.occurrence_datetime,
  s.party_id,
	ir.immunization_reaction_id,
	ir.reaction_date,
	ir.manifestation_code,
	ir.manifestation_concept_id,
	ir.manifestation_raw_code,
	ir.manifestation_raw_concept_id,
	ir.manifestation_observation_id,
	ir.reported AS reaction_reported,
	ir.rank AS reaction_rank,
	ir.period_start AS reaction_period_start,
	ir.period_end AS reaction_period_end,
	h.period_start AS hub_period_start,
	h.period_end AS hub_period_end,
	s.period_start,
	s.period_end,
	s.load_datetime,
	s.record_source_id,
	s.data_source_code,
	s.data_source_concept_id,
	s.data_source_raw_code,
	s.data_source_raw_concept_id

    FROM budd.qdp_immunizations.hub_immunization h
    JOIN budd.qdp_immunizations.sat_immunization s
      ON h.immunization_id = s.immunization_id
    JOIN budd.qdp_immunizations.sat_immunization_reaction ir
      ON h.immunization_id = ir.immunization_id
    JOIN budd.qdp_refcodes.tennant t
      on h.tennant_id = t.tennant_id
;

SELECT * from view_immunization_reactions;


-- -----------------------------------------------------------
-- create view_immunization_protocols view
-- -----------------------------------------------------------
CREATE OR REPLACE VIEW view_immunization_protocols AS
SELECT
  h.tennant_id,
  t.tennant_name,
  h.immunization_id,
	h.immunization_entity_id,
  s.vaccine_code,
  s.vaccine_concept_id,
  s.status_code,
  s.occurrence_datetime,
  s.party_id,
	pa.immunization_protocol_applied_id,
	pa.series,
	pa.authority_organisation_id,
	pa.target_disease_code,
	pa.target_disease_concept_id,
	pa.target_disease_raw_code,
	pa.target_disease_raw_concept_id,
	pa.dose_number,
	pa.series_doses,
	pa.rank AS protocol_rank,
	pa.period_start AS protocol_period_start,
	pa.period_end AS protocol_period_end,
	h.period_start AS hub_period_start,
	h.period_end AS hub_period_end,
	s.period_start,
	s.period_end,
	s.load_datetime,
	s.record_source_id,
	s.data_source_code,
	s.data_source_concept_id,
	s.data_source_raw_code,
	s.data_source_raw_concept_id

    FROM budd.qdp_immunizations.hub_immunization h
    JOIN budd.qdp_immunizations.sat_immunization s
      ON h.immunization_id = s.immunization_id
    JOIN budd.qdp_immunizations.sat_immunization_protocol_applied pa
      ON h.immunization_id = pa.immunization_id
    JOIN budd.qdp_refcodes.tennant t
      on h.tennant_id = t.tennant_id
;

SELECT * from view_immunization_protocols;


/**

**/
```
